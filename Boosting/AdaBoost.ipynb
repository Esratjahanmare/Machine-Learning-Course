```python
## importing models
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB 
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import  AdaBoostClassifier
import pandas as  pd

from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

import warnings
warnings.filterwarnings("ignore")
import os
```

```python
df=pd.read_csv('heart.csv')
rows=df.shape[0]
columns=df.shape[1]

print ( rows,columns )
## input Variables = x
## output variables = y 

y=df['target']

x=df.drop('target',axis=1)
x
```

```python
## input Variables = x
## output variables = y 

y=df['target']

x=df.drop('target',axis=1)
x
```

```python
## test and Train Split

X_train,X_test,y_train, y_test = train_test_split( x,y,test_size=.25,random_state=42323232)
```

```python
log_clf = LogisticRegression()
rnd_clf = RandomForestClassifier()
svm_clf = SVC()
nbg_clf = GaussianNB()
```

```python
for clf in (log_clf, nbg_clf, rnd_clf, svm_clf):
    clf.fit(X_train, y_train)
    
    y_pred = clf.predict(X_test)
    
    print(clf.__class__.__name__, accuracy_score(y_test, y_pred))
```

```python
from sklearn.svm import SVC
svc_clf= SVC( probability =True,kernel='linear')
```

### AdaBoostClassifier with SVM

```python
boost = AdaBoostClassifier( estimator=svc_clf, n_estimators=500,algorithm='SAMME',learning_rate=.5)

boost.fit(X_train, y_train)

y_pred = boost.predict(X_test)

print("Bagging with Random Forest Classifier :Accuracy ", accuracy_score(y_test, y_pred))
```

```python
## Training Score

boost.score(X_train,y_train)
```

```python
## Testing Score

boost.score(X_test,y_test)
```

### AdaBoostClassifier with Decision Tree

```python
boost = AdaBoostClassifier( estimator=DecisionTreeClassifier(max_depth=1), n_estimators=10,algorithm='SAMME',learning_rate=0.5)

boost.fit(X_train, y_train)

y_pred = boost.predict(X_test)

print("Bagging with Decision Tree Classifier :Accuracy ", accuracy_score(y_test, y_pred))
```

```python
## Training Score

boost.score(X_train,y_train)
```

```python
## Testing Score

boost.score(X_test,y_test)
```
